# Lectura del *DataFrame* para entrenamiento

```python
import pandas as pd
df = pd.read_csv('dataset_train.csv')
df.head()
```

# Se hace un recuento de sus archivos Null

```python
df.info()
df.isnull().sum()
```

# Importación de *train_test_split* para dividir los valores de entrenamiento y prueba

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'] + ' ' + df['emotion'], df['sentimento'], test_size=0.10)
```

# Importación del vectorizador

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizador = CountVectorizer()
X_train_vec = vectorizador.fit_transform(X_train)
X_test_vec = vectorizador.transform(X_test)
print(len(df['emotion']), len(df['text']), len(df['sentimento']))
```

# Importación del modelo *LogisticRegression*

```python
from sklearn.linear_model import LogisticRegression

modelo = LogisticRegression()
modelo.fit(X_train_vec, y_train)
```

# Lectura del *DataFrame* de prueba

```python
df_test = pd.read_csv('dataset_test_sin_sentimento.csv')
```

# Contabilización de sus datos Null

```python
df_test.info()
df_test.isnull().sum()
```

# Creación del archivo de prueba

```python
nueva = df_test['text'] + ' ' + df_test['emotion']
X_nueva = vectorizador.transform(nueva)
predict = modelo.predict(X_nueva)

df_test['etiqueta'] = predict

output = pd.DataFrame({'ID': df_test['id'], 'etiqueta': predict})
output.to_csv('submission3.csv', index=False)
```
